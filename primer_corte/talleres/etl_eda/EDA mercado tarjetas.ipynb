Análisis exploratorio de datos sobre el registro de ventas generado en la actividad de tarjetas de producto.

**Dataset:** cada fila es una venta entre grupos, con columnas `id_venta, vendedor, comprador, producto, categoria, precio_lista, precio_venta, cantidad, calificacion, ronda`.

## Paso 0 · Cargar los datos
Si ya tienes el CSV con los datos reales de tu grupo, súbelo con la celda de abajo. Si no, se usa el dataset de ejemplo (con los mismos errores intencionales de la guía).

```python
import pandas as pd

# Si tienes tu propio CSV de la actividad, descomenta estas 3 líneas para subirlo:
# from google.colab import files
# subido = files.upload()
# df = pd.read_csv(list(subido.keys())[0])

# Dataset de ejemplo (mismo de la guía, con errores intencionales)
datos = {
    'id_venta': [1, 2, 3, 4, 5, 6],
    'vendedor': ['Grupo A', 'Grupo B', 'Grupo C', 'Grupo A', 'Grupo B', 'Grupo C'],
    'comprador': ['Grupo C', 'Grupo A', 'Grupo B', 'Grupo B', 'Grupo C', 'Grupo A'],
    'producto': ['Termo Acero 500ml', 'Audífonos BT Mini', 'Cuaderno Ecológico',
                 'Termo Acero 500ml', 'Powerbank 10000mAh', 'Cuaderno Ecológico'],
    'categoria': ['Hogar', 'Tecnología', 'Papelería', 'Hogar', 'Tecnología', 'Papelería'],
    'precio_lista': ['25000', '60000', '8000', '25.000', '45000', '8000'],
    'precio_venta': ['22000', '65000', '8000', '20000', None, '8000'],
    'cantidad': [2, 1, 3, 1, 1, 3],
    'calificacion': [4, 5, 3, 4, 4, 3],
    'ronda': [1, 1, 1, 2, 2, 1],
}
df = pd.DataFrame(datos)
df
```

## Paso 2 · Calidad de datos
❯ ¿Qué venta no podríamos usar sin corregirla primero?

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df['precio_venta'] = (df['precio_venta']
    .astype(str)
    .str.replace('.', '', regex=False)
    .str.replace(',', '.', regex=False))
df['precio_venta'] = pd.to_numeric(df['precio_venta'], errors='coerce')

df = df.drop_duplicates()
df
```

## Paso 3 · Análisis univariado
❯ ¿Alguna venta tiene un precio que no encaja con el resto?

```python
df['precio_venta'].describe()
```

```python
df['categoria'].value_counts().plot(kind='bar', title='Ventas por categoría')
```

```python
df.boxplot(column='precio_venta')
```

## Paso 4 · Análisis bivariado
❯ ¿Qué categoría vende más caro y qué vendedor generó más ingresos?

```python
df.plot.scatter(x='cantidad', y='precio_venta', title='Cantidad vs precio de venta')
```

```python
df.groupby('categoria')['precio_venta'].mean().plot(kind='bar', title='Precio promedio por categoría')
```

```python
df.groupby('vendedor')['precio_venta'].sum().sort_values(ascending=False)
```

## Paso 5 · Correlación
❯ ¿Los productos mejor calificados se vendieron más caros?

```python
df[['precio_venta', 'calificacion']].corr()
```

```python
df.plot.scatter(x='calificacion', y='precio_venta', title='Calificación vs precio de venta')
```
